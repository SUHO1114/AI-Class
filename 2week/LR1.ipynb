```python
import pandas as pd  # 데이터를 표(Table) 형태로 다루기 위한 Pandas 라이브러리를 불러오기.
import numpy as np   # 수치 계산 및 행렬 연산을 위한 NumPy 라이브러리를 불러오기

# 데이터 파일이 저장된 경로를 문자열 변수 'path'에 저장.
path = '/content/drive/MyDrive/Colab Notebooks/diabetes.csv'

# pd.read_csv() 함수를 사용하여 해당 경로의 CSV 파일을 읽어오기.
# 읽어온 데이터는 Pandas의 핵심 자료구조인 'DataFrame(데이터프레임)' 형태로 변수 'df'에 저장.
df = pd.read_csv(path)

df
```

```python
# 전체 데이터프레임(df)에서 'Outcome' 열(Column)만 추출하여 변수 'y'에 저장.
y = df['Outcome']

# 추출된 'y' Series 내에 각 클래스(0과 1)가 몇 개씩 들어있는지 개수를 세어 출력.
y.value_counts()

y
```

```python
# 전체 데이터프레임(df)에서 'Outcome' 열만 제거(drop)하여 변수 'X'에 저장.
# axis=1은 '열(Column)'을 기준으로 삭제하겠다는 의미. (axis=0은 행 삭제)
X = df.drop('Outcome', axis=1)

# 분리된 데이터프레임 'X'의 상위 5개 행(Row)만 출력하여 구조를 확인.
X.head()

X
```

```python
# Scikit-learn 라이브러리의 model_selection 모듈에서 데이터를 나누는 함수를 불러오기.
from sklearn.model_selection import train_test_split

# 전체 데이터(X, y)를 특정 비율에 따라 4개의 덩어리로 쪼개기.
# 1. X_train: 모델 학습에 사용할 입력 데이터 (문제지)
# 2. X_test:  모델 성능을 평가할 때 사용할 입력 데이터 (시험지)
# 3. y_train: X_train에 대응하는 실제 정답 (문제지의 해설지)
# 4. y_test:  X_test에 대응하는 실제 정답 (시험지의 정답지)
X_train, X_test, y_train, y_test = train_test_split(
    X,                # 독립 변수 (Features)
    y,                # 종속 변수 (Target)
    test_size=0.25,   # 전체 데이터의 25%를 테스트(검증)용으로 사용하고, 나머지 75%를 학습용으로 사용.
    random_state=0    # 데이터를 섞을 때 사용하는 '랜덤 시드' 값.
                      # 숫자를 고정하면 코드를 다시 실행해도 똑같은 데이터 분할 결과를 얻음.
)
```

```python
# 머신러닝 알고리즘 및 평가 지표 라이브러리를 불러오기.
from sklearn.linear_model import LogisticRegression  # 이진 분류에 주로 사용되는 로지스틱 회귀 모델
from sklearn.svm import SVC                           # 서포트 벡터 머신 (Support Vector Classifier)
from sklearn.neural_network import MLPClassifier      # 인공신경망 (Multi-layer Perceptron)
from sklearn.tree import DecisionTreeClassifier       # 의사결정나무 (Decision Tree)
from sklearn.metrics import accuracy_score            # 정확도 측정 함수
from sklearn.metrics import confusion_matrix          # 오차 행렬 (예측 오류의 종류 확인)

# 1. 로지스틱 회귀 모델 객체를 생성.
# random_state=0으로 설정하여 실행할 때마다 결과가 동일하게 나오도록 고정.
clf_lr = LogisticRegression(random_state=0)

# 2. 모델을 학습. (Train)
# 학습 데이터(X_train)와 그에 맞는 정답(y_train)을 모델에게 주어 관계를 파악하게 함.
clf_lr.fit(X_train, y_train)

# 3. 학습된 모델을 사용하여 테스트 데이터(X_test)에 대한 예측을 수행.
# 모델은 정답(y_test)을 보지 못한 상태에서 오직 X_test 데이터만 보고 예측값을 내놓는다.
pred_lr = clf_lr.predict(X_test)

# 4. 모델의 성능을 출력.
print ("\n--- Logistic Regression Classifier ---")

# 정확도(Accuracy): 전체 예측 중 맞춘 비율을 계산. (0~1 사이 값)
print (accuracy_score(y_test, pred_lr))

# 오차 행렬(Confusion Matrix): 모델이 정답을 어떻게 틀렸는지 세부 정보를 보여줌.
# [[TN, FP],
#  [FN, TP]] 형태로 출력되며, 어느 클래스를 헷갈려하는지 파악할 때 유용.
print (confusion_matrix(y_test, pred_lr))
# TN (True Negative) = 116: 실제로 정상인 사람을 모델이 정상이라고 올바르게 맞힌 경우.
# FP (False Positive) = 14: 실제로는 정상인데 모델이 당뇨라고 잘못 예측한 경우.
# FN (False Negative) = 25: 실제로는 당뇨 환자인데 모델이 정상이라고 잘못 예측한 경우.
# TP (True Positive) = 37: 실제 당뇨 환자를 모델이 당뇨라고 올바르게 맞힌 경우.
```

```python
# 1. 의사결정나무(Decision Tree) 모델 객체를 생성.
# random_state=0을 사용하여 트리가 만들어지는 과정의 무작위성을 고정.
clf_dt = DecisionTreeClassifier(random_state=0)

# 2. 모델을 학습. (Train)
# X_train(입력 변수)과 y_train(정답) 사이의 규칙(Rule)을 학습하여 트리 구조를 형성.
clf_dt.fit(X_train, y_train)

# 3. 학습된 트리 모델을 사용하여 테스트 데이터(X_test)에 대한 예측을 수행.
pred_dt = clf_dt.predict(X_test)

# 4. 모델의 성능을 출력.
print ("\n--- Decision Tree Classifier ---")

# 정확도(Accuracy): 의사결정나무가 맞춘 전체 비율을 계산.
print (accuracy_score(y_test, pred_dt))

# 오차 행렬(Confusion Matrix):
print (confusion_matrix(y_test, pred_dt))
```

```python
# Scikit-learn의 ensemble 모듈에서 RandomForestClassifier를 불러오기.
# 'Ensemble'은 여러 모델을 합쳐 성능을 높이는 기법을 의미.
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import pandas as pd
import warnings

# 불필요한 경고 메시지(버전 업데이트 관련 등)가 출력되지 않도록 설정.
warnings.filterwarnings('ignore')

print ("\n--- Random Forest ---")

# 1. 랜덤 포레스트 모델 객체를 생성. (기본적으로 100개의 나무를 만듦.)
# random_state=0으로 설정하여 결과가 매번 바뀌지 않도록 고정.
rf_clf = RandomForestClassifier(random_state=0)

# 2. 모델을 학습. (Train)
# X_train(데이터)과 y_train(정답)을 모델에 넣어 나무 100개를 학습.
rf_clf.fit(X_train, y_train)

# 3. 학습된 앙상블 모델을 사용하여 테스트 데이터(X_test)의 결과를 예측.
pred = rf_clf.predict(X_test)

# 4. 모델 성능 지표를 출력.
# 정확도(Accuracy): 전체 중 맞춘 비율
print(accuracy_score(y_test, pred))

# 오차 행렬(Confusion Matrix): 모델이 0과 1을 각각 어떻게 맞히거나 틀렸는지 확인.
print(confusion_matrix(y_test, pred))
```
