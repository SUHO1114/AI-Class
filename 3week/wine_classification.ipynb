# **# 데이터 세트 불러오기**
`import pandas as pd` : 데이터를 표(Table) 형태로 다루기 위한 **Pandas** 라이브러리.  
`import numpy as np` : 수치 계산 및 행렬 연산을 위한 **NumPy** 라이브러리.  
`path = '/content/drive/MyDrive/Colab Notebooks/3week/wine.csv'` : 데이터 파일이 저장된 경로를 문자열 변수 '**path**'에 저장.  
`df = pd.read_csv(path)` : 읽어온 데이터는 데이터프레임 형태로 변수 '**df**'에 저장.  








```python
import pandas as pd
import numpy as np
path = '/content/drive/MyDrive/Colab Notebooks/3week/wine.csv'
df = pd.read_csv(path)

df
```

# **# 데이터 전처리(y)**
`y = df['Wine']` : 예측하고자 하는 목표 변수(와인 종류)를 변수 '**y**'에 저장.  
`y.value_counts()` : 각 와인 클래스별 데이터 개수를 확인.

```python
y = df['Wine']
y.value_counts()
```

# **# 데이터 전처리(X)**
`X = df.drop('Wine', axis=1)` : 전체 데이터에서 '**Wine**' 열을 제외한 나머지 특징들을 변수 '**X**'에 저장.  
`X.head()` : 분리된 특징 데이터의 상위 5개 행을 확인.

```python
X=df.drop('Wine',axis=1)
X.head()
```

# **# 데이터셋 분리 (Training & Test Set)**
`from sklearn.model_selection import train_test_split` : 데이터를 나누기 위한 사이킷런의 함수.  
`test_size=0.25` : 전체 데이터의 **25%** 를 평가용(Test)으로, **75%** 를 학습용(Train)으로 설정.  
`random_state=0` : 실행할 때마다 동일한 결과가 나오도록 난수 시드 고정.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.25,random_state=0)
```

# **# 모델별 학습 및 평가 1 - 로지스틱 회귀 (Logistic Regression)**
**선형 회귀 방식을 분류에 적용한 알고리즘으로, 클래스에 속할 확률을 계산.**  
`clf_lr = LogisticRegression(random_state=0)` : 1. **로지스틱 회귀** 모델 객체 생성.  
`clf_lr.fit(X_train, y_train)` : 2. 학습 데이터(**X_train**)와 그에 맞는 정답(**y_train**)을 모델에게 주어 관계를 파악하는 방식으로 학습.    
`pred_lr = clf_lr.predict(X_test)` : 3. 학습된 모델을 사용하여 테스트 데이터(**X_test**)에 대한 예측을 수행.  
`print (accuracy_score(y_test, pred_lr))` : 4. 전체 예측 중 맞춘 비율인 정확도(**Accuracy**) 출력. (0~1 사이 값).     
`print (confusion_matrix(y_test, pred_lr))` : 5. 모델이 정답을 어떻게 틀렸는지 세부 정보를 보여주는 오차 행렬(**Confusion Matrix**) 출력.  

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

clf_lr = LogisticRegression(random_state=0)
clf_lr.fit(X_train, y_train)

pred_lr = clf_lr.predict(X_test)

print ("\n--- Logistic Regression Classifier ---")
print (accuracy_score(y_test, pred_lr))
print (confusion_matrix(y_test, pred_lr))
```

# **# 모델별 학습 및 평가 2 - 결정 트리 (Decision Tree)**
**데이터의 특징을 기준으로 '예/아니오' 질문을 던져 데이터를 분류하는 나무 형태의 모델.**  
`clf_dt = DecisionTreeClassifier(random_state=0)` : 1. **의사결정나무** 모델 객체를 생성.  
`clf_dt.fit(X_train, y_train)` : 2. 학습 데이터(**X_train**)와 정답(**y_train**) 사이의 규칙을 학습하여 트리 구조를 형성하는 방식.    
`pred_dt = clf_dt.predict(X_test)` : 3. 학습된 모델을 사용하여 테스트 데이터(**X_test**)에 대한 예측을 수행.  
`print (accuracy_score(y_test, pred_dt))` : 4. 전체 예측 중 맞춘 비율인 정확도(**Accuracy**) 출력. (0~1 사이 값).     
`print (confusion_matrix(y_test, pred_dt))` : 5. 모델이 정답을 어떻게 틀렸는지 세부 정보를 보여주는 오차 행렬(**Confusion Matrix**) 출력.  

```python
clf_dt = DecisionTreeClassifier(random_state=0)
clf_dt.fit(X_train, y_train)

pred_dt = clf_dt.predict(X_test)

print ("\n--- Decision Tree Classifier ---")
print (accuracy_score(y_test, pred_dt))
print (confusion_matrix(y_test, pred_dt))
```

# **# 모델별 학습 및 평가 3 - 랜덤 포레스트 (Random Forest)**
**여러 개의 결정 트리를 생성하고, 그 결과들을 종합하여 예측 성능을 높이는 방식.**  
`clf_dt = DecisionTreeClassifier(random_state=0)` : 1. **랜덤 포레스트** 모델 객체를 생성.  
`clf_dt.fit(X_train, y_train)` : 2. 학습 데이터(**X_train**)와 정답(**y_train**) 모델에 넣어 나무를 만드는 방식으로 학습.    
`pred_dt = clf_dt.predict(X_test)` : 3. 학습된 모델을 사용하여 테스트 데이터(**X_test**)에 대한 예측을 수행.  
`print (accuracy_score(y_test, pred_dt))` : 4. 전체 예측 중 맞춘 비율인 정확도(**Accuracy**) 출력. (0~1 사이 값).     
`print (confusion_matrix(y_test, pred_dt))` : 5. 모델이 정답을 어떻게 틀렸는지 세부 정보를 보여주는 오차 행렬(**Confusion Matrix**) 출력.  

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

rf_clf = RandomForestClassifier(random_state=0)
rf_clf.fit(X_train, y_train)

pred = rf_clf.predict(X_test)

print ("\n--- Random Forest ---")
print(accuracy_score(y_test,pred))
print (confusion_matrix(y_test, pred))
```

# **# 모델별 학습 및 평가 4 - kNN 알고리즘 (k-Nearest Neighbor)**
**새로운 데이터가 들어왔을 때, 가장 가까운 $k$개의 이웃 데이터를 확인하여 더 많은 쪽으로 분류하는 방식.**  
`knn = KNeighborsClassifier(n_neighbors=6)` : 1. 새로운 데이터를 분류할 때 참고할 이웃의 개수를 **6개**로 설정.  
`knn.fit(X_train, y_train)` : 2. 학습 데이터(**X_train**)와 정답(**y_train**)을 사용하여 모델을 학습.  
`pred_y = knn.predict(X_test)` : 3. 학습된 모델을 사용하여 테스트 데이터(**X_test**)에 대한 예측을 수행.  
`accuracy_score(y_test, pred_y)` : 4. 전체 예측 중 맞춘 비율인 정확도(**Accuracy**) 출력. (0~1 사이 값).     
`print (confusion_matrix(y_test, pred_y))` : 5. 모델이 정답을 어떻게 틀렸는지 세부 정보를 보여주는 오차 행렬(**Confusion Matrix**) 출력.   

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics

knn = KNeighborsClassifier(n_neighbors=6)
knn.fit(X_train, y_train)

pred_y = knn.predict(X_test)

print ("\n--- kNN ---")
print (accuracy_score(y_test,pred_y))
print (confusion_matrix(y_test, pred_y))
```
