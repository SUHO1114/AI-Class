```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

path = '/content/drive/MyDrive/Colab Notebooks/breast_cancer.csv'

# DataFrame으로 열기 df = data frame
df2 = pd.read_csv(path)

df2
```

```python
y = df2['worst compactness']
y.value_counts()
y
```

```python
X=df2.drop('worst compactness',axis=1)
X.head()
X
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.25,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

model = LinearRegression()
model.fit(X_train, y_train)
ly_preds = model.predict(X_test)

print('평균제곱근오차', mean_squared_error(ly_preds, y_test))

def mse_np(actual, predicted):
    return np.mean((np.array(actual) - np.array(predicted)) ** 2)

print('평균제곱근오차', mse_np(ly_preds, y_test))

def mse(actual, predicted):
    sum_square_error = sum((a - p) ** 2 for a, p in zip(actual, predicted))
    mean_square_error = sum_square_error / len(actual)
    return mean_square_error

print('평균제곱근오차', mse(ly_preds, y_test))
```

```python
plt.figure(figsize=(10,5))
plt.scatter(X_test['mean compactness'], y_test, label='y_test')
plt.scatter(X_test['mean compactness'], ly_preds, c='y', label='ly_preds')
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor()
model.fit(X_train, y_train)

from sklearn.metrics import mean_squared_error

dy_preds = model.predict(X_test)
print('평균제곱근오차', mean_squared_error(dy_preds, y_test))
```

```python
plt.figure(figsize=(10,5))
plt.scatter(X_test['mean compactness'], y_test, label='y_test')
plt.scatter(X_test['mean compactness'], dy_preds, c='g', label='dy_preds')
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()
model.fit(X_train, y_train)

from sklearn.metrics import mean_squared_error

ry_preds = model.predict(X_test)
print('평균제곱근오차', mean_squared_error(ry_preds, y_test))
```

```python
plt.figure(figsize=(10,5))
plt.scatter(X_test['mean compactness'], y_test, label='y_test')
plt.scatter(X_test['mean compactness'], ry_preds, c='orange', label='ry_preds')

plt.show()
```
